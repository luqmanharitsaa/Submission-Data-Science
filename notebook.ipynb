## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

2. **Pencilan (Outliers):** Berdasarkan analisis sebaran data (`describe`), terdapat nilai pencilan (*outliers*) ekstrem pada kolom `casual` di `day_df` yang melampaui batas atas kuartil (Q3 + 1.5 * IQR).

- Kolom dteday masih bertipe object sehingga belum bisa dimanfaatkan secara optimal untuk ekstraksi waktu atau agregasi tanggal tanpa dikonversi ke datetime.

### Cleaning Data

#### Fixing data problem

```python
# --- MASALAH 1: Memperbaiki Tipe Data dteday & Mapping Kategori ---
day_df['dteday'] = pd.to_datetime(day_df['dteday'])
hour_df['dteday'] = pd.to_datetime(hour_df['dteday'])

# Mapping musim dan cuaca agar informatif
season_map = {1: 'Spring', 2: 'Summer', 3: 'Fall', 4: 'Winter'}
weather_map = {
    1: 'Clear/Partly Cloudy',
    2: 'Misty/Cloudy',
    3: 'Light Snow/Rain',
    4: 'Severe Weather'
}

day_df['season'] = day_df['season'].map(season_map)
day_df['weathersit'] = day_df['weathersit'].map(weather_map)

hour_df['season'] = hour_df['season'].map(season_map)
hour_df['weathersit'] = hour_df['weathersit'].map(weather_map)


# --- MASALAH 2: Menangani Outlier pada Kolom casual (day_df) Menggunakan Metode IQR ---
# Catat jumlah baris sebelum pembersihan
baris_sebelum = len(day_df)

# Hitung Q1, Q3, dan IQR
Q1 = day_df['casual'].quantile(0.25)
Q3 = day_df['casual'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

# Filter data untuk membuang outlier
day_df = day_df[(day_df['casual'] >= lower_bound) & (day_df['casual'] <= upper_bound)]

# Catat jumlah baris sesudah pembersihan
baris_sesudah = len(day_df)

# Tampilkan jumlah baris sebelum dan sesudah sesuai kriteria reviewer
print(f"Jumlah baris day_df sebelum dibersihkan : {baris_sebelum}")
print(f"Jumlah baris day_df sesudah dibersihkan : {baris_sesudah}")
print(f"Total baris pencilan yang dibersihkan   : {baris_sebelum - baris_sesudah}")

# Ekspor data bersih untuk kebutuhan dashboard
hour_df.to_csv('dashboard/main_data.csv', index=False)
```

**Insight:** (Opsional)
- Kolom dteday kini sudah berformat tanggal baku (datetime64[ns]), sehingga siap digunakan untuk analisis berbasis deret waktu.  
- Kolom kategorikal (season, weathersit, yr) menjadi jauh lebih intuitif dan mudah diinterpretasikan untuk tahap Exploratory Data Analysis (EDA) dan visualisasi tanpa perlu mengingat kode angka lagi.

## Exploratory Data Analysis (EDA)

### Explore
Menghitung rata-rata, nilai maksimum, nilai minimum, dan total akumulasi peminjaman sepeda berdasarkan kategori musim (season) dan kondisi cuaca (weathersit) pada day_df.   Menghitung rata-rata peminjaman sepeda per jam (hr) berdasarkan status hari kerja (workingday) pada hour_df

```python
# 1. Eksplorasi day_df: Rata-rata peminjaman berdasarkan musim
season_analysis = day_df.groupby('season')['cnt'].agg(['mean', 'max', 'min', 'sum']).reset_index()
season_analysis = season_analysis.sort_values(by='mean', ascending=False)
print("--- Statistik Peminjaman Berdasarkan Musim ---")
display(season_analysis)

# 2. Eksplorasi day_df: Rata-rata peminjaman berdasarkan kondisi cuaca
weather_analysis = day_df.groupby('weathersit')['cnt'].agg(['mean', 'max', 'min', 'sum']).reset_index()
weather_analysis = weather_analysis.sort_values(by='mean', ascending=False)
print("\n--- Statistik Peminjaman Berdasarkan Cuaca ---")
display(weather_analysis)

# 3. Eksplorasi hour_df: Rata-rata peminjaman per jam berdasarkan tipe hari (Hari Kerja vs Libur)
# workingday: 1 = Hari Kerja, 0 = Akhir Pekan / Hari Libur
hourly_pattern = hour_df.groupby(['workingday', 'hr'])['cnt'].mean().reset_index()
print("\n--- Cuplikan Pola Peminjaman per Jam ---")
display(hourly_pattern.head(10))
```

**Insight:** (Opsional)
- Pengaruh Musim: Musim gugur (Fall) mencatat rata-rata peminjaman harian tertinggi (\~5.644 unit/hari), diikuti musim panas (Summer) dan musim dingin (Winter). Musim semi (Spring) memiliki rata-rata peminjaman paling rendah (~2.604 unit/hari).

- Pengaruh Cuaca: Kondisi cuaca cerah (Clear) menghasilkan rata-rata peminjaman harian tertinggi (~4.877 unit/hari), sedangkan kondisi hujan ringan/salju (Light Snow/Rain) menurunkan aktivitas peminjaman secara drastis hingga rata-rata ~1.803 unit/hari.

- Pola Jam Sibuk: Pada hari kerja (workingday = 1), peminjaman melonjak tajam pada jam berangkat kerja (pukul 08.00) dan jam pulang kerja (pukul 17.00–18.00). Sebaliknya, pada hari libur (workingday = 0), aktivitas peminjaman terdistribusi merata dari siang hingga sore hari (pukul 12.00–16.00).

## Visualization & Explanatory Analysis

### Pertanyaan 1: Bagaimana pengaruh kondisi cuaca (weathersit) dan musim (season) terhadap total rata-rata peminjaman sepeda harian sepanjang tahun 2011 hingga 2012?


```python
# Menyiapkan canvas dengan 2 subplot (kiri: Musim, kanan: Cuaca)
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(16, 6))

# Palet warna konsisten
colors_season = ["#72BCD4", "#D3D3D3", "#D3D3D3", "#D3D3D3"]
colors_weather = ["#72BCD4", "#D3D3D3", "#D3D3D3"]

# 1. Barplot Pengaruh Musim
season_order = day_df.groupby('season')['cnt'].mean().sort_values(ascending=False).index
sns.barplot(
    x='season',
    y='cnt',
    data=day_df,
    order=season_order,
    palette=colors_season,
    ax=ax[0],
    errorbar=None
)
ax[0].set_title('Rata-rata Peminjaman Sepeda Berdasarkan Musim', fontsize=14, weight='bold')
ax[0].set_xlabel(None)
ax[0].set_ylabel('Rata-rata Peminjaman Harian (Unit)', fontsize=12)

# 2. Barplot Pengaruh Cuaca
weather_order = day_df.groupby('weathersit')['cnt'].mean().sort_values(ascending=False).index
sns.barplot(
    x='weathersit',
    y='cnt',
    data=day_df,
    order=weather_order,
    palette=colors_weather,
    ax=ax[1],
    errorbar=None
)
ax[1].set_title('Rata-rata Peminjaman Sepeda Berdasarkan Kondisi Cuaca', fontsize=14, weight='bold')
ax[1].set_xlabel(None)
ax[1].set_ylabel(None)

plt.tight_layout()
plt.show()
```

### Pertanyaan 2: Bagaimana pola perbedaan jumlah peminjaman sepeda per jam antara hari kerja (workingday) dan hari libur (weekend/holiday) pada tahun 2011–2012?

```python
# Menghitung rata-rata peminjaman per jam berdasarkan tipe hari
hourly_trend = hour_df.groupby(['hr', 'workingday'])['cnt'].mean().reset_index()

# Mapping label workingday untuk legend grafik
hourly_trend['workingday'] = hourly_trend['workingday'].map({1: 'Hari Kerja (Workingday)', 0: 'Hari Libur / Weekend'})

# Membuat visualisasi pola jam
plt.figure(figsize=(12, 6))
sns.lineplot(
    data=hourly_trend,
    x='hr',
    y='cnt',
    hue='workingday',
    palette=['#FF6F61', '#1F77B4'],
    linewidth=2.5,
    marker='o'
)

plt.title('Pola Rata-rata Peminjaman Sepeda per Jam (Hari Kerja vs Hari Libur)', fontsize=14, weight='bold')
plt.xlabel('Jam (00:00 - 23:00)', fontsize=12)
plt.ylabel('Rata-rata Jumlah Peminjaman (Unit)', fontsize=12)
plt.xticks(range(0, 24))
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='Kategori Hari', title_fontsize='11', fontsize='10')

plt.tight_layout()
plt.show()
```

**Insight:** (Opsional)
- Insight Pertanyaan 1:
Musim gugur (Fall) dan cuaca cerah (Clear) menjadi pendorong utama volume peminjaman harian tertinggi. Peminjaman anjlok drastis saat terjadi hujan ringan/salju (Light Snow/Rain).   
- Insight Pertanyaan 2:
Peminjaman di hari kerja memiliki dua lonjakan tajam pada jam sibuk (08.00 pagi dan 17.00–18.00 sore). Sedangkan di hari libur, peminjaman terdistribusi stabil pada siang hari hingga sore (12.00–16.00).

## Analisis Lanjutan (Opsional)

```python
# 1. Denormalisasi suhu menjadi skala derajat Celsius (rentang asli berkisar hingga 41°C)
day_df['temp_celsius'] = day_df['temp'] * 41

# 2. Binning kategori suhu lingkungan
day_df['temp_category'] = pd.cut(
    day_df['temp_celsius'],
    bins=[-float('inf'), 15, 25, float('inf')],
    labels=['Dingin (<15°C)', 'Sejuk (15-25°C)', 'Hangat/Panas (>25°C)']
)

# 3. Binning tingkat permintaan peminjaman sepeda (Low, Medium, High Demand)
day_df['demand_category'] = pd.qcut(
    day_df['cnt'],
    q=3,
    labels=['Low Demand', 'Medium Demand', 'High Demand']
)

# 4. Agregasi statistik peminjaman berdasarkan klaster suhu
temp_cluster = day_df.groupby('temp_category')['cnt'].agg(
    total_hari='count',
    rata_rata='mean',
    total_peminjaman='sum'
).reset_index()

print("--- Hasil Pengelompokan Berdasarkan Kategori Suhu ---")
display(temp_cluster)

# 5. Visualisasi distribusi tingkat permintaan pada tiap kategori suhu
plt.figure(figsize=(10, 5))
sns.countplot(
    data=day_df,
    x='temp_category',
    hue='demand_category',
    palette=['#E57373', '#FFD54F', '#81C784']
)
plt.title('Distribusi Tingkat Permintaan Sepeda Berdasarkan Kategori Suhu', fontsize=14, weight='bold')
plt.xlabel('Kategori Suhu', fontsize=12)
plt.ylabel('Jumlah Hari', fontsize=12)
plt.legend(title='Tingkat Permintaan', frameon=True)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

**Insight:** (Opsional)

- Dominasi Suhu Rendah pada Low Demand: Pada kategori suhu dingin (<15°C), mayoritas hari (162 dari 221 hari) berada pada tingkat peminjaman rendah (Low Demand) dengan rata-rata hanya ~2.833 unit/hari.

- Perpindahan ke High Demand pada Suhu Nyaman: Ketika suhu naik ke rentang sejuk (15–25°C) dan hangat (>25°C), rata-rata peminjaman harian melonjak drastis masing-masing menjadi ~4.864 unit/hari dan ~5.632 unit/hari, didominasi oleh kategori Medium dan High Demand.

- Implikasi Operasional: Fluktuasi suhu harian dapat dijadikan indikator utama dalam peramalan kapasitas armada, di mana kebutuhan unit di lapangan berlipat ganda saat suhu berada di atas 15°C.

## Conclusion & Recommendation

- **Conclusion pertanyaan 1:**
- Kondisi cuaca dan musim memiliki pengaruh yang sangat signifikan terhadap volume peminjaman sepeda.
- Faktor Musim: Rata-rata peminjaman sepeda harian tertinggi terjadi pada musim gugur (Fall) dengan rata-rata ~5.644 unit/hari, disusul oleh musim panas (Summer) ~4.992 unit/hari dan musim dingin (Winter) ~4.728 unit/hari. Musim semi (Spring) mencatatkan aktivitas terendah dengan rata-rata ~2.604 unit/hari.
- Faktor Cuaca: Pengguna sepeda sangat sensitif terhadap presipitasi dan jarak pandang. Hari dengan cuaca cerah (Clear) menghasilkan peminjaman tertinggi (\~4.877 unit/hari), cuaca berkabut/mendung (Misty/Cloudy) mengalami sedikit penurunan (~4.036 unit/hari), sedangkan kondisi hujan ringan atau salju (Light Snow/Rain) menurunkan peminjaman secara drastis hingga rata-rata ~1.803 unit/hari (turun lebih dari 60% dibanding hari cerah).

- **Conclusion pertanyaan 2:**
- Terdapat perbedaan pola penggunaan sepeda yang sangat kontras antara hari kerja (workingday) dan hari libur (weekend/holiday).   
- Hari Kerja (Workingday): Grafik peminjaman membentuk pola bimodal (dua puncak tajam) yang bertepatan dengan jam sibuk mobilitas pekerja/pelajar, yaitu puncak pagi pada pukul 08.00 (\~477 unit/jam) dan puncak sore pada pukul 17.00–18.00 (~492–525 unit/jam). Aktivitas di luar jam tersebut relatif rendah.
- Hari Libur (Weekend/Holiday): Grafik peminjaman membentuk pola unimodal (lonjakan kurva tunggal yang landai) khas aktivitas rekreasi, di mana peminjaman mulai meningkat sejak pagi, mencapai titik tertinggi pada siang hingga sore hari (pukul 12.00–16.00 dengan volume stabil ~350–370 unit/jam), lalu perlahan menurun menjelang malam.

**Rekomendasi Action Item:**
- Optimalisasi Distribusi Unit Sepeda (Rebalancing): Pada hari kerja, tim operasional harus memprioritaskan redistribusi unit sepeda ke stasiun di sekitar area pemukiman/stasiun transit sebelum pukul 07.00 pagi, dan menarik kembali unit dari kawasan perkantoran/pusat bisnis sebelum pukul 16.30 sore.

- Manajemen Pemeliharaan (Maintenance Fleet): Jadwalkan perbaikan rutin dan perawatan armada pada musim semi (Spring) atau hari-hari dengan prakiraan cuaca hujan lebat, saat demand pasar sedang berada pada titik terendah agar tidak mengganggu ketersediaan unit saat peak season.

- Strategi Promosi & Insentif Harga: Buat program langganan khusus komuter (commuter pass) untuk mendongkrak loyalitas pengguna harian pada hari kerja, serta tawarkan tarif promo siang hari (leisure weekend discount) untuk menarik lebih banyak pengguna kasual di akhir pekan.
